```python
%matplotlib widget
import numpy as np
import json
import matplotlib.pyplot as plt
from pydae.plot_tools import set_style_web
colors = set_style_web(plt)
```

```python
with open('../record.json', 'r') as fobj:
    data_dict = json.load(fobj)
```

```python
t = np.array([item['t'] for item in data_dict])
q_ref_pu = np.array([item['q_ref_pu'] for item in data_dict])
Q_POI = np.array([item['Q_POI'] for item in data_dict])
q_ppc = np.array([item['q_ppc'] for item in data_dict])

p_ref_pu = np.array([item['p_ref_pu'] for item in data_dict])
P_POI = np.array([item['P_POI'] for item in data_dict])
p_ppc = np.array([item['p_ppc'] for item in data_dict])
```

```python
fig, axes = plt.subplots()
axes.plot(t,q_ref_pu, label='q_ref_pu')
axes.plot(t,Q_POI/2e6, label='Q_POI')
#axes.plot(t,q_ppc, label='q_ppc')
axes.legend()
```

```python
fig, axes = plt.subplots()
axes.plot(np.diff(t))
```

```python
fig, axes = plt.subplots()
axes.plot(t,p_ref_pu)
axes.plot(t,P_POI/2e6)
axes.plot(t,p_ppc)
```
