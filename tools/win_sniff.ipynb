```python
from scapy.all import sniff

def process_packet(packet):
    # Display basic packet information
    print(f"Packet: {packet.summary()}")
    
    # Check for IP packets and print source/destination
    if packet.haslayer("IP"):
        ip_layer = packet["IP"]
        print(f"Source: {ip_layer.src} -> Destination: {ip_layer.dst}")
        print("-" * 50)

# Sniff packets on the LAN interface (default: eth0 or wlan0 on Linux)
# Use iface parameter to specify a specific network interface
print("Starting packet sniffing... Press Ctrl+C to stop.")
sniff(filter="ip", prn=process_packet, store=0)
```

```python
import pyshark

def monitor_traffic(interface="eth0"):
    capture = pyshark.LiveCapture(interface=interface)
    print(f"Listening on {interface}... Press Ctrl+C to stop.")
    for packet in capture.sniff_continuously():
        try:
            print(f"Packet: {packet}")
            if hasattr(packet, 'ip'):
                print(f"Source: {packet.ip.src} -> Destination: {packet.ip.dst}")
        except AttributeError:
            pass  # Handle packets without IP layer

# Specify your interface (e.g., wlan0, eth0)
monitor_traffic(interface="Wireless LAN adapter Wi-Fi")
```

```python
import pyshark
import asyncio

async def monitor_traffic(interface="wlan0"):
    capture = pyshark.LiveCapture(interface=interface)
    print(f"Listening on {interface}... Press Ctrl+C to stop.")
    async for packet in capture.sniff_continuously():
        try:
            print(f"Packet: {packet}")
        except AttributeError:
            pass

if __name__ == "__main__":
    asyncio.run(monitor_traffic())
```
